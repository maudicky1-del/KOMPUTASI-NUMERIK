```python

# =====================================================
# PROGRAM PENYELESAIAN SPLTV
# METODE ELIMINASI GAUSS DENGAN PARTIAL PIVOTING
# =====================================================

import numpy as np

print("========================================")
print(" PROGRAM PENYELESAIAN SPLTV")
print(" METODE ELIMINASI GAUSS")
print("========================================")

# 1. MENENTUKAN PERSAMAAN SPLTV
# 2X + Y + Z = 10
# X + 2Y + Z = 11
# X + Y + 2Z = 12

A = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
], dtype=float)

B = np.array([10, 11, 12], dtype=float)

# 2. MEMBENTUK MATRIKS AUGMENTED
M = np.column_stack((A, B))

print("\nMatriks Augmented Awal:")
print(M)

# 3. PROSES ELIMINASI GAUSS
n = 3
tol = 1e-10

for i in range(n):

    # Partial Pivoting
    pivot = i + np.argmax(np.abs(M[i:, i]))

    # Menukar baris
    if pivot != i:
        M[[i, pivot]] = M[[pivot, i]]

    # Eliminasi elemen di bawah pivot
    for j in range(i + 1, n):
        faktor = M[j, i] / M[i, i]
        M[j, i:] -= faktor * M[i, i:]

    print(f"\nMatriks setelah eliminasi ke-{i+1}:")
    print(M)

# 4. SUBSTITUSI BALIK
X = np.zeros(n)

for i in range(n - 1, -1, -1):
    X[i] = (
        M[i, n] - np.dot(M[i, i+1:n], X[i+1:n])
    ) / M[i, i]

# 5. MENAMPILKAN HASIL
print("\n========================================")
print(" HASIL PERHITUNGAN SPLTV")
print("========================================")

print(f"Nilai X = {X[0]:.2f}")
print(f"Nilai Y = {X[1]:.2f}")
print(f"Nilai Z = {X[2]:.2f}")

# 6. VERIFIKASI
print("\n========================================")
print(" VERIFIKASI HASIL")
print("========================================")

hasil = A @ X

for i in range(n):
    print(f"Persamaan {i+1}: {hasil[i]:.2f} = {B[i]:.2f}")

print("\nProgram selesai dijalankan.")
```
