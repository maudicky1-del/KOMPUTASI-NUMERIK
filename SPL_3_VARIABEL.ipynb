```python
# ============================================================
# PROGRAM PENYELESAIAN SISTEM PERSAMAAN LINEAR 3 VARIABEL
# (SPLTV) DENGAN METODE ELIMINASI GAUSS
# Komputasi Numerik - Google Colab
# ============================================================

import numpy as np

def spltv_gauss():
    print("=" * 55)
    print(" PROGRAM PENYELESAIAN SPLTV 3 VARIABEL")
    print(" METODE ELIMINASI GAUSS")
    print("=" * 55)

    # 1. Input koefisien dan konstanta
    A = []
    b = []

    print("\nMasukkan koefisien untuk setiap persamaan:")
    for i in range(3):
        print(f"\nPersamaan ke-{i + 1}:")
        x = float(input("Koefisien X: "))
        y = float(input("Koefisien Y: "))
        z = float(input("Koefisien Z: "))
        konstanta = float(input("Konstanta: "))
        A.append([x, y, z])
        b.append(konstanta)

    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    # 2. Membentuk matriks augmented
    M = np.column_stack((A, b))
    print("\nMatriks augmented:")
    print(M)

    # 3. Eliminasi Gauss dengan pivot parsial
    n = 3
    for i in range(n):
        pivot = i + np.argmax(np.abs(M[i:, i]))
        M[[i, pivot]] = M[[pivot, i]]

        if abs(M[i, i]) < 1e-10:
            break

        for j in range(i + 1, n):
            faktor = M[j, i] / M[i, i]
            M[j, :] = M[j, :] - faktor * M[i, :]

    print("\nMatriks setelah eliminasi Gauss:")
    print(M)

    # 4. Memeriksa apakah sistem konsisten
    for i in range(n):
        if np.all(np.abs(M[i, :3]) < 1e-10) and abs(M[i, 3]) >= 1e-10:
            print("\nKesimpulan: sistem tidak memiliki solusi.")
            return

    # 5. Substitusi mundur
    if any(abs(M[i, i]) < 1e-10 for i in range(n)):
        print("\nKesimpulan: sistem memiliki solusi tak hingga.")
        print("Program ini menampilkan solusi tunggal saja.")
        return

    hasil = np.zeros(n)
    for i in range(n - 1, -1, -1):
        jumlah = np.dot(M[i, i + 1:n], hasil[i + 1:n])
        hasil[i] = (M[i, 3] - jumlah) / M[i, i]

    # 6. Menampilkan hasil perhitungan
    print("\n" + "=" * 55)
    print("HASIL PERHITUNGAN")
    print("=" * 55)
    print(f"Nilai X = {hasil[0]:.4f}")
    print(f"Nilai Y = {hasil[1]:.4f}")
    print(f"Nilai Z = {hasil[2]:.4f}")

    # 7. Verifikasi hasil
    print("\nVerifikasi hasil ke persamaan awal:")
    hasil_ruas_kiri = A @ hasil
    for i in range(n):
        print(f"Persamaan {i + 1}: {hasil_ruas_kiri[i]:.4f} = {b[i]:.4f}")

    print("\nKesimpulan: nilai X, Y, dan Z berhasil dihitung.")
    print("=" * 55)


# Menjalankan program
spltv_gauss()
```
